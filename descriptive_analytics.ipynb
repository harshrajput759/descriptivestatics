Q1: Understanding Central Tendency (Easy)
A bakery tracks the daily sales of muffins (in dozens) over a week: [10, 12, 11, 15, 14, 13, 12].
What is the most representative value of their weekly sales, and why?

Ans. The most representative value of the bakery’s weekly muffin sales is 12 dozens per day, as it is both the median and the mode, making it a reliable measure of central tendency for this dataset.

Q4: Median in Real Life (Medium)
A car dealer notes the prices of used cars: [$8,000, $9,500, $10,200, $11,000, $50,000].
Why is the median a better measure than the mean in this case? Calculate the median.

Ans. The prices are already in ascending order:

8,000,
9,500
10,200.
11,000,
50,000

Median=10,200

The median price of the used cars is $10,200. In this case, the median is a better measure than the mean because the dataset contains an extreme value ($50,000) that would distort the average, while the median represents the typical car price more accurately.

A student times how long it takes to finish a puzzle each day: [25, 30, 27, 35, 40].

Ans. Range=Maximum−Minimum

Range=40−25=15 minutes

The range of the puzzle-solving times is 15 minutes. This shows that there is a noticeable variation in the student’s performance across different days, with completion times spread over a 15-minute interval.
